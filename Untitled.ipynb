```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Food.csv")
order_time_count = df["Order_Time"].value_counts()
plt.figure(figsize=(10,6))
plt.bar(order_time_count.index, order_time_count.values)
plt.title("Orders by Order Time")
plt.xlabel("Order Time")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Food.csv")
popular_dishes = df["Item_Name"].value_counts()
plt.figure(figsize=(10,6))
plt.plot(popular_dishes.index, popular_dishes.values, marker='o')
plt.title("Most Popular Dishes")
plt.xlabel("Dish Name")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
